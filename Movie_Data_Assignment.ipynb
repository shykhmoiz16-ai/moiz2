### **Question 29**
Har movie ke 'Genre' column ko split karke check karein ki total unique genres pure dataset mein kitne hain.

```python
all_genres = []
for genre in data.Genre:
    for item in genre.split(", "):
        all_genres.append(item)
print("Total unique genres:", len(set(all_genres)))
print(set(all_genres))
```

### **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
data.fillna({"Certificate": data.Certificate.mode()[0],
             "Meta_score": data.Meta_score.mean(),
             "Gross": data.Gross.median()}, inplace=True)

data.to_csv("cleaned_movies.csv", index=False)
print("File saved as cleaned_movies.csv")
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
drama_gross = data[data.Genre.str.contains("Drama")]["Gross"].dropna()
action_gross = data[data.Genre.str.contains("Action")]["Gross"].dropna()

plt.boxplot([drama_gross, action_gross], tick_labels=["Drama", "Action"])
plt.title("Gross Revenue: Drama vs Action")
plt.ylabel("Gross Revenue")
plt.show()
```
